```python
import numpy as np
import pandas as pd

np.random.seed(7)

marks = np.random.normal(65, 12, (500, 6))
marks = np.clip(marks, 0, 100)

print("Dataset Shape:", marks.shape)
print("\nFirst Five Students:\n", marks[:5])
```

```python
student_sum = np.sum(marks, axis=1)
student_mean = np.mean(marks, axis=1)
student_median = np.median(marks, axis=1)
student_std = np.std(marks, axis=1)
student_variance = np.var(marks, axis=1)

subject_mean = np.mean(marks, axis=0)
subject_median = np.median(marks, axis=0)
subject_std = np.std(marks, axis=0)
subject_variance = np.var(marks, axis=0)

print("Student Average:", student_mean[:5])
print("Student Median:", student_median[:5])
print("Student Standard Deviation:", student_std[:5])
print("Student Variance:", student_variance[:5])

print("\nSubject Average:", subject_mean)
print("Subject Median:", subject_median)
print("Subject Standard Deviation:", subject_std)
print("Subject Variance:", subject_variance)
```

```python
z_scores = (
    marks - np.mean(marks, axis=0)
) / np.std(marks, axis=0)

outlier_indices = np.where(np.abs(z_scores) > 2)

print("Total Outliers:", len(outlier_indices[0]))

for student, subject in zip(*outlier_indices):
    print(
        f"Student {student + 1}, "
        f"Subject {subject + 1}, "
        f"Marks: {marks[student, subject]:.2f}"
    )
```

```python
student_rank = np.argsort(-student_sum)

print("TOP 10 STUDENTS\n")

for rank, student in enumerate(student_rank[:10], start=1):
    print(
        f"Rank {rank}: Student {student + 1} "
        f"| Total Marks: {student_sum[student]:.2f}"
    )
```

```python
final_data = np.hstack((
    marks,
    student_sum.reshape(-1, 1)
))

print("Final Dataset Shape:", final_data.shape)
print("\nFirst Five Records:\n", final_data[:5])
```

```python
columns = [
    "Maths",
    "Physics",
    "Chemistry",
    "Computer Science",
    "English",
    "Biology",
    "Total Marks"
]

df = pd.DataFrame(final_data, columns=columns)

df.insert(0, "Student ID", range(1, 501))

df["Rank"] = df["Total Marks"].rank(
    ascending=False,
    method="min"
).astype(int)

df = df.sort_values("Rank")

display(df.head(10))

df.to_csv("student_performance.csv", index=False)

print("CSV exported successfully!")
```
